```python
# 1. Remove exact duplicate rows
df = df.drop_duplicates()

# 2. Remove rows with unusable date/price/country
df = df.dropna(subset=["InvoiceDate", "Price", "Country"])

# 3. Fill missing descriptions where StockCode has a known description
description_mode = (
    df[df["Description"].notna()]
    .groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0])
)

df["Description"] = df["Description"].fillna(
    df["StockCode"].map(description_mode)
)

# 4. Recalculate transaction type and revenue
df["TransactionType"] = np.where(
    df["Quantity"] < 0, "Return", "Sale"
)

df["Revenue"] = df["Quantity"] * df["Price"]
```

### Querying DataFrame with SQL using `pandasql`

Since you're working with pandas DataFrames, you can treat them like SQL tables using the `pandasql` library. First, we'll install the library, then import it, and finally demonstrate how to run a SQL query on your `df2` DataFrame.

```python
from pandasql import sqldf

# Define a simple SQL query. The DataFrame name 'df2' is used as the table name.
query = """
SELECT
    Country,
    COUNT(DISTINCT `Customer ID`) AS unique_customers,
    SUM(Revenue) AS total_revenue
FROM
    df2
GROUP BY
    Country
ORDER BY
    total_revenue DESC
LIMIT 10;
"""

# Execute the SQL query on the DataFrame
result_df = sqldf(query, globals())

# Display the results
display(result_df)
```

```python
from pandasql import sqldf

def run_sql_query(sql_query_string, dataframe_vars=globals()):
    """
    Executes a SQL query against the DataFrames available in dataframe_vars
    and displays the result.
    """
    print(f"Executing query:\n{sql_query_string}")
    result = sqldf(sql_query_string, dataframe_vars)
    display(result)
    return result
```

Now, you can use this `run_sql_query` function. For example, to get the sum of revenue again, or to find the average quantity per country:

```python
# Example 1: Get the sum of Revenue using the helper function
query_sum_revenue = """SELECT sum(Revenue) FROM df2"""
run_sql_query(query_sum_revenue)

# Example 2: Find the average Quantity per Country
query_avg_quantity = """
SELECT
    Country,
    AVG(Quantity) AS average_quantity
FROM
    df2
GROUP BY
    Country
ORDER BY
    average_quantity DESC
LIMIT 5;
"""
run_sql_query(query_avg_quantity)
```
